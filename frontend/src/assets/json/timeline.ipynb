Creating json from csv

```python
import pandas as pd

# Load the CSV files
df_section_1 = pd.read_csv("dura_language_books_materials_research.csv")
df_section_2 = pd.read_csv("dura_language_campaigns.csv")
df_section_3 = pd.read_csv("dura_language_classes.csv")

# Convert each DataFrame to a JSON file
df_section_1.to_json(
    "dura_language_books_materials_research.json",
    orient="records",
    force_ascii=False,
    lines=False,
)
df_section_2.to_json(
    "dura_language_campaigns.json", orient="records", force_ascii=False, lines=False
)
df_section_3.to_json(
    "dura_language_classes.json", orient="records", force_ascii=False, lines=False
)

print("JSON files created successfully.")
```
